>> #### метрики бинарной классификации: ROC AUC(метрика ранжирования)
оценивает качество ранжировани нулей и единиц между собой
вероятность того, что два случайно взятых из выборки объекта разных классов будут упорядочены моделью корректно, то есть уверенность модели на объекте класса 1 будет выше, чем на объекте класса 0

## метрики многоклассовой классификации: микро макро усреднение
микро: для каждого класса метрику, среднее арифметическое
макро: ?
если классы сбаналсированы, неважно
если нет, то макро, если важнны все классы, и микро, если важен каждый класс

>> #### почему f1-score - среднее гармоническое?
не хотим, чтобы precision: 1, recall: 0
среднее гармоническое будет равно нулю, тогда как среднее арифметическое будет равно 0.5
срелднее геометрическое обладает этими же свойства гармонического, но мнее пессиместично

>> #### если precision большой, recall маленький, а нужно наоборот, что делаеть?
понизить порог отсечения вероятности
