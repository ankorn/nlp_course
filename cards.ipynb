>> #### tricky questions from interview: nlp
- почему ллм галлюцинирует? лосс поощряет угадывание
- почему контекстное окно маленькое? квадратичная сложность аттеншена
- what is the diff .detach, .no_grad? .detach: comp graph discarded; .no_grad: no comp graph
- how to build knowledge graph? NER, relation extraction('invented'), entity linking(wiki)

- why rnn grad van? packprop though time multiplies derivatives actross time; lots of W_hh mults, so van, expl; use grad clipping
- lstm: hidden[-1] == out[:,-1,:] always? no, when bidaractional embs are concatinated
- lstm: use full hidden for: nmt, pos tagging, ner

- tf-idf: d_freq * all_ds_freq_penalty; "the" - low, "clusterisation" - high; don't know meaning, scales badly; advanced tf-idf: bm25(rag comp)
- dim of text vecorized via tf-idf? voc_size
- why log in c-e? max likelihood estimation: ∏ p_{i, true_class}; -> 0, so hard to differentiate; max log(p) == min -log(p) == min c-e

- how many parameters lora learns? two mat: proj to rank and proj to emb_dim, so 2 * rank * emb_dim
- why residual? fix grad van to stack many layers; fix losing pos enb signal; fix over-smoothing: after many attns token embs converge to single vector
- how to extend vocab? add tokens to vocab, fine-tune model on new data(same task: mlm, nsp for bert)
- bert inf: pooler_output: [CLS] -> proj; pure [CLS]: last_hidden_state[:, 0, :] # 0-th token
- why decoder input seq shifted right? for model to not see t_i when predict t_i
- why sin, cos pos embs? represent relative positions via linear transformations

- attn: padding mask: extended attention_mask
- 1/sqrt(d_k)? softmax smoothing => meaningful gradients(not too small)
- attn: Q @ K.T = X @ Q_w @ K_w.T @ X.T; why not train single A = (Q_w @ K_w.T)? not possible in cross-attn
- what if A in decoder(self-attn)? possible, even better memory; but brakes K, V cache
- why mha? heads in parallel => better time, memory same; also interpretability

- BN: [b * s, emb_dim], over feats(columns), so emb_dim means, vars, params; LN: means and vars of emb(still emb_dim params, shared across embs)
- почему нет больших екодеров? почему начали скейлить декожеры? утилизация данных 100% в дек, а в енк(mlm) предсказывается 15% не маскированных токенов

- не помещается на gpu? DataParallel(model instance on every gpu, grands on batch, avg over all instances) не подходит; ModelParallel(e.g. pipeline parallelism: layer-wize: every layer on gpu)
- во время обучения что хранится на gpu? X, W, Adam(2 for each model param), c graph
- mix precision: weights fp32; forward and backward fp16
- как представлены флоаты? бит знака, мантисса(точность), экспонента(рейндж)
- что важнее во время форварда, что во время бекварда? forward: norm, so range less imp; softmax, so precision imp; backward: large, small grad, so range
- flash attn: main issue is memory, so 1) attn matmul block by block on fast SRAM; 2) not store full attn scores during forward, just store softmax statistics and recompute attn scores during backwards
- moe(from dence to sparse): ffn => set of ffn; trained router chooses which to use(small part of them); model keeps quality, but has lower amount of active parameters, so fast inf
- speculative decoding(inference): generate with assistant model, verify in single forward pass of main model; get exactly the same result, as if infering main
- continuous batching: gpu utilization; dynamicly remove finished gens, add new; address padding issue: e.g. group by seq len
